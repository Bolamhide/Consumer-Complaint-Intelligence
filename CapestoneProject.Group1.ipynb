```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train_tfidf, y_train)
```

```python
y_pred = model.predict(X_test_tfidf)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

## Chapter 4 — Results, Interpretation and Business Application

### 4.1 Model Results

The baseline Logistic Regression model achieved an accuracy of 82% on the test set. However, the classification report showed that the model correctly identified the majority class, Debt collection, but did not correctly identify the minority class, Credit reporting, credit repair services, or other personal consumer reports. The macro F1-score of 0.45 indicates that the model's performance was uneven across the two complaint categories.

To address the observed class imbalance, a second Logistic Regression model was trained using balanced class weights. The balanced model achieved an accuracy of 78%. Although this approach gave greater consideration to the minority class during training, it did not produce correct predictions for the minority class on the test set. This indicates that class weighting alone was not sufficient to overcome the limitations of the available data.

### 4.2 Interpretation of Model Performance

The model results show that overall accuracy alone does not provide a complete picture of classification performance. Although the baseline model achieved 82% accuracy, its confusion matrix showed that all 7 minority-class complaints were classified as Debt collection. The balanced model produced a lower accuracy of 78% and also failed to correctly classify the minority class.

This result is strongly influenced by the distribution of the dataset. Debt collection accounted for 166 of the 199 records used for modelling, while Credit reporting, credit repair services, or other personal consumer reports accounted for only 33 records. The imbalance made it difficult for the model to learn sufficient patterns for the minority category.

Therefore, the model should not be considered reliable enough for fully automated complaint categorization. Its current performance is more appropriate for decision support, where predictions can be reviewed by a human before complaints are routed or acted upon.

### 4.3 Confidence and Threshold Policy

A confidence threshold was introduced to determine when the model should make an automatic classification and when a complaint should be referred for human review. A threshold of 70% was selected as the minimum confidence level for accepting a prediction.

The results showed that only 5% of the test complaints reached or exceeded the 70% confidence threshold, while 95% were classified as "Needs Review". The average prediction confidence was approximately 61.4%, with a maximum confidence of approximately 70.9%.

These results indicate that the model is not sufficiently confident to support widespread automated complaint routing. The threshold policy therefore provides a safety mechanism by directing low-confidence predictions to human reviewers rather than allowing uncertain classifications to be used automatically.

### 4.4 Business Application

The proposed system can support financial service organisations by automatically analysing the text of consumer complaints and suggesting the most likely complaint category. This can help reduce the time required to manually review and route complaints.

However, based on the model results, the system should currently be used as a decision-support tool rather than a fully automated routing system. High-confidence predictions may be considered for preliminary categorisation, while low-confidence cases should be referred to human reviewers.

This approach can help organisations improve complaint-handling workflows while maintaining human oversight for uncertain cases.

### 4.5 Data Critique and Limitations

The dataset presented several limitations that affected the classification results. The main limitation was class imbalance. Debt collection represented 166 records, while Credit reporting, credit repair services, or other personal consumer reports represented only 33 records. A separate Credit reporting category contained only one record and was removed from the modelling dataset because there was insufficient information to train and evaluate that category reliably.

The dataset used for modelling was also relatively small, containing 199 records after the cleaning and category-selection steps. This limited the amount of information available for learning patterns from complaint narratives.

Another limitation was the model's inability to correctly identify the minority class in the test set. Although class weighting was applied, the balanced model still recorded zero recall for the minority category. This suggests that additional data, particularly more representative examples of minority complaint categories, would be required to improve the system.

The 70% confidence threshold also resulted in 95% of test cases being referred for review. This indicates that the current model does not provide sufficient confidence for extensive automated routing and should therefore be treated as a decision-support prototype rather than a production-ready automated system.

### 4.6 Practical Application

In a practical financial services environment, the system could receive a consumer complaint narrative as input and use the trained machine-learning model to suggest a complaint category. The prediction could then support the initial routing of complaints to the appropriate department.

Because the current model has limited performance on the minority category and low prediction confidence for most test cases, human review should remain part of the workflow. Complaints with confidence below the defined threshold should be flagged for manual assessment.

With a larger and more balanced dataset, the system could be retrained and evaluated again to determine whether it is suitable for wider operational use.

## Chapter 5 — Conclusion

### 5.1 Summary of Findings

This project investigated the use of machine learning for automated classification of financial services consumer complaints using complaint narratives from the CFPB Consumer Complaint Database. The study applied data cleaning, exploratory analysis, TF-IDF text representation, and Logistic Regression to classify complaints into product categories.

The baseline model achieved 82% accuracy, but its performance was strongly affected by class imbalance. A balanced Logistic Regression model was also evaluated and achieved 78% accuracy, but it did not correctly classify the minority category.

A confidence-based threshold policy was introduced to support safer practical use of the model. At a 70% confidence threshold, only 5% of the test complaints were considered sufficiently confident for automatic classification, while 95% were referred for human review.

Overall, the findings show that the approach can support complaint categorization, but the current dataset and model are not sufficiently reliable for fully automated complaint routing.

### 5.2 Conclusion

The project demonstrates how machine learning and natural language processing techniques can be applied to financial services consumer complaints. The use of TF-IDF and Logistic Regression provided a simple and explainable approach for converting complaint narratives into a form that could be used for classification.

The results also demonstrated the importance of evaluating class distribution, rather than relying on accuracy alone. The strong imbalance in the dataset limited the model's ability to identify the minority complaint category. The confidence analysis further showed that the model requires human oversight for most predictions.

Therefore, the developed system should be viewed as a prototype decision-support solution. Its performance and practical usefulness could be improved through the collection of a larger and more balanced dataset and further model development.

### 5.3 Recommendations

The following recommendations are made based on the findings of the project:

1. A larger and more balanced complaint dataset should be collected to improve the model's ability to learn minority complaint categories.

2. Future model development should continue to evaluate performance using precision, recall, F1-score, and confusion matrices rather than accuracy alone.

3. Human review should remain part of the complaint-routing process when model confidence is below the defined threshold.

4. The model should be retrained periodically using new complaint data to reflect changes in complaint patterns and financial service issues.

5. Further testing with additional machine-learning and natural language processing techniques may be considered when a larger dataset becomes available.

### 5.4 Future Considerations

Future development of the system should focus on improving the quality and size of the dataset used for training. A larger dataset with better representation of different complaint categories would provide more reliable training and evaluation results.

Future work could also investigate other classification techniques and compare their performance with the Logistic Regression approach used in this project. The confidence threshold could be further evaluated and adjusted based on the level of risk associated with incorrect complaint classification.

Before practical deployment, the system should be tested with a larger and more representative dataset. Human oversight should also remain available for complaints that the model cannot classify with sufficient confidence.
